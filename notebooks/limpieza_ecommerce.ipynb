# **Limpieza del Dataset**

---

## 🔹 Paso 1: Cargar y revisar el dataset

```python
import pandas as pd

df = pd.read_csv("../data/ecommerce_dataset_updated.csv")

# Ver primeras filas
print(df.head())

# Ver info general
print(df.info())

# Ver valores nulos
print(df.isnull().sum())
```

----
## 🔹 Paso 2: Renombrar columnas
Evitamos espacios y símbolos para que coincidan con la tabla SQL:


```python
df = df.rename(columns={
    "Price (Rs.)": "Price",
    "Discount (%)": "Discount",
    "Final_Price(Rs.)": "Final_Price"
})
```

----
## 🔹 Paso 3: Limpiar valores
- **Fechas**: convertir a formato estándar `YYYY-MM-DD`.
- **Categorías y métodos de pago**: normalizar mayúsculas/minúsculas.
- **Descuentos y precios**: asegurar que sean numéricos.


```python
# Convertir fecha
df["Purchase_Date"] = pd.to_datetime(df["Purchase_Date"], dayfirst=True)

# Normalizar texto
df["Category"] = df["Category"].str.strip().str.title()
df["Payment_Method"] = df["Payment_Method"].str.strip().str.title()

# Asegurar numéricos
df["Price"] = pd.to_numeric(df["Price"], errors="coerce")
df["Discount"] = pd.to_numeric(df["Discount"], errors="coerce")
df["Final_Price"] = pd.to_numeric(df["Final_Price"], errors="coerce")
```

---

## 🔹 Paso 4: Manejo de valores nulos y duplicados

```python
# Rellenar nulos en descuento con 0
df["Discount"] = df["Discount"].fillna(0)

# Rellenar nulos en método de pago con 'Unknown'
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
```

---

## 🔹 Paso 5: Guardar dataset limpio

```python
df.to_csv("../data/ecommerce_dataset_clean.csv", index=False)
```

----
## 🧹 Proceso de Limpieza del Dataset

Antes de cargar el dataset en la base de datos SQLite, se realizó un proceso de limpieza para asegurar consistencia y calidad en los datos.

### 1. Renombrar columnas
Los nombres originales contenían espacios y símbolos (`Price (Rs.)`, `Discount (%)`, `Final_Price(Rs.)`).  
Se renombraron a nombres más simples y compatibles con SQL:
- `Price`
- `Discount`
- `Final_Price`

### 2. Conversión de tipos
- **Fechas**: convertidas al formato estándar `YYYY-MM-DD` para facilitar consultas temporales.  
- **Precios y descuentos**: convertidos a valores numéricos (`REAL`).  
- **Categorías y métodos de pago**: normalizados en texto (`TEXT`), con capitalización uniforme.

### 3. Manejo de valores nulos
- `Discount`: valores nulos reemplazados por `0`.  
- `Payment_Method`: valores nulos reemplazados por `"Unknown"`.  

### 4. Eliminación de duplicados
Se eliminaron registros duplicados para evitar sesgos en los cálculos de KPIs.

### 5. Exportación del dataset limpio
El dataset resultante se guardó como `ecommerce_dataset_clean.csv`, listo para ser cargado en la base de datos SQLite.
